## 3. Binomial Distribution

In this section, I used the binomial distribution to look at how many houses might cost more than USD 450,000 in a sample of 10 randomly selected houses.

For this analysis, a house is counted as a success if its price is above USD 450,000. I used the housing dataset to work out the probability of this happening.

```python
# Calculate the probability of success

p = (df['price'] > 450000).mean()

print("Probability of success:", p)
```

```python
from scipy.stats import binom

# Number of houses selected
n = 10

# Possible number of successes
x = np.arange(0, n + 1)

# Calculate binomial probabilities
pmf_values = binom.pmf(x, n=n, p=p)

print(pmf_values)
```

```python
# Plot the binomial distribution

plt.bar(x, pmf_values)

plt.xlabel('Number of houses above $450,000')
plt.ylabel('Probability')
plt.title('Binomial Distribution of House Prices')

plt.show()
```

### Interpretation of the Binomial Distribution

Looking at the graph, we can see the possible number of houses costing more than USD 450,000 out of a sample of 10 houses.

Based on the dataset, the probability of a house costing more than USD 450,000 is about 49.73%. This means that a house has almost an equal chance of being above or at or below this price.

The graph is nearly symmetrical because the probability is close to 50%. The highest point is at 5 houses, with a probability of around 24.61%.

Overall, we would expect about 5 out of 10 randomly selected houses to cost more than USD 450,000. Results close to 5 are more likely, while getting 0 or all 10 houses above this price is much less likely.

```python
# Binomial distribution for a larger sample

n_large = 100

x_large = np.arange(0, n_large + 1)

pmf_large = binom.pmf(x_large, n=n_large, p=p)

plt.bar(x_large, pmf_large)

plt.xlabel('Number of houses above $450,000')
plt.ylabel('Probability')
plt.title('Binomial Distribution for 100 Houses')

plt.show()
```

### Interpretation: Effect of Increasing the Sample Size

When I increased the sample size from 10 to 100 houses, I noticed that the binomial distribution became smoother and looked more like a bell-shaped curve.

For a sample of 100 houses, we would expect around 50 houses to cost more than USD 450,000, based on the probability of 49.73% calculated from the dataset.

The graph is also almost symmetrical because the probability is very close to 50%. This shows that when the sample size is large enough, the binomial distribution starts to look similar to a normal distribution.

Although the number of houses above USD 450,000 can vary more in a larger sample, the percentage tends to stay closer to 49.73%. This makes the results more consistent when looking at proportions rather than the actual number of houses.
