## 3. Binomial Distribution

In this section, I used the binomial distribution to look at how many houses might cost more than USD 450,000 in a sample of 10 randomly selected houses.

For this analysis, a house is counted as a success if its price is above USD 450,000. I used the housing dataset to work out the probability of this happening.

```python
# Calculate the probability of success

p = (df['price'] > 450000).mean()

print("Probability of success:", p)
```

```python
from scipy.stats import binom

# Number of houses selected
n = 10

# Possible number of successes
x = np.arange(0, n + 1)

# Calculate binomial probabilities
pmf_values = binom.pmf(x, n=n, p=p)

print(pmf_values)
```

```python
# Plot the binomial distribution

plt.bar(x, pmf_values)

plt.xlabel('Number of houses above $450,000')
plt.ylabel('Probability')
plt.title('Binomial Distribution of House Prices')

plt.show()
```

### Interpretation of the Binomial Distribution

Looking at the graph, we can see the possible number of houses costing more than USD 450,000 out of a sample of 10 houses.

Based on the dataset, the probability of a house costing more than USD 450,000 is about 49.73%. This means that a house has almost an equal chance of being above or at or below this price.

The graph is nearly symmetrical because the probability is close to 50%. The highest point is at 5 houses, with a probability of around 24.61%.

Overall, we would expect about 5 out of 10 randomly selected houses to cost more than USD 450,000. Results close to 5 are more likely, while getting 0 or all 10 houses above this price is much less likely.

```python
# Binomial distribution for a larger sample

n_large = 100

x_large = np.arange(0, n_large + 1)

pmf_large = binom.pmf(x_large, n=n_large, p=p)

plt.bar(x_large, pmf_large)

plt.xlabel('Number of houses above $450,000')
plt.ylabel('Probability')
plt.title('Binomial Distribution for 100 Houses')

plt.show()
```

### Interpretation: Effect of Increasing the Sample Size

When I increased the sample size from 10 to 100 houses, I noticed that the binomial distribution became smoother and looked more like a bell-shaped curve.

For a sample of 100 houses, we would expect around 50 houses to cost more than USD 450,000, based on the probability of 49.73% calculated from the dataset.

The graph is also almost symmetrical because the probability is very close to 50%. This shows that when the sample size is large enough, the binomial distribution starts to look similar to a normal distribution.

Although the number of houses above USD 450,000 can vary more in a larger sample, the percentage tends to stay closer to 49.73%. This makes the results more consistent when looking at proportions rather than the actual number of houses.

## 4. Poisson Distribution

In this section, I used the Poisson distribution to look at how often houses with seven or more bedrooms might appear in a sample.

I chose houses with seven or more bedrooms because they are quite rare in the dataset.

Using the information from the housing dataset, I calculated the average rate and used it to estimate how many of these houses we might find in a sample of 100 houses.

```python
# Calculate the proportion of houses with 7 or more bedrooms

p_rare = (df['bedrooms'] >= 7).mean()

# Expected number of rare houses in a sample of 100
n = 100

lam = n * p_rare

print("Probability of a rare house:", p_rare)
print("Average number of rare houses:", lam)

```

```python
from scipy.stats import poisson

# Possible number of rare houses
x_poisson = np.arange(0, 6)

# Calculate Poisson probabilities
pmf_poisson = poisson.pmf(x_poisson, mu=lam)

print(pmf_poisson)
```

```python
# Plot the Poisson distribution

plt.bar(x_poisson, pmf_poisson)

plt.xlabel('Number of houses with 7 or more bedrooms')
plt.ylabel('Probability')
plt.title('Poisson Distribution of Rare Houses')

plt.show()
```

### Interpretation of the Poisson Distribution

Looking at the graph, we can see how likely it is to find houses with seven or more bedrooms in a sample of 100 houses.

From the dataset, the chance of a house having seven or more bedrooms is only about 0.287%. This means we would expect around 0.287 such houses in a sample of 100, which is less than one house on average.

The graph is strongly skewed to the right. There is about a 75.06% chance of finding no houses with seven or more bedrooms, while the chance of finding exactly one is around 21.53%.

These results show just how uncommon houses with seven or more bedrooms are in the dataset. The Poisson distribution is useful here because it helps us estimate how often these rare houses might appear in a random sample, assuming each selection is independent and the probability stays the same.

```python
# Poisson distribution for a larger sample

n_large = 1000

lam_large = n_large * p_rare

x_large = np.arange(0, 12)

pmf_large = poisson.pmf(x_large, mu=lam_large)

plt.bar(x_large, pmf_large)

plt.xlabel('Number of houses with 7 or more bedrooms')
plt.ylabel('Probability')
plt.title('Poisson Distribution for 1000 Houses')

plt.show()
```

### Interpretation: Effect of Increasing the Sample Size on the Poisson Distribution

When I increased the sample size from 100 to 1,000 houses, I noticed a clear difference in the Poisson distribution. The expected number of houses with seven or more bedrooms increased from about 0.287 to 2.869.

For a sample of 100 houses, there is a 75.06% chance of finding none with seven or more bedrooms. However, when the sample increases to 1,000 houses, the most likely result is finding two such houses, while the chance of finding none drops to around 5.68%.

The graph also changes shape. With 100 houses, most of the probability is concentrated at zero, but with 1,000 houses, the distribution spreads out and becomes less skewed to the right.

This shows that even though houses with seven or more bedrooms are still rare, we are more likely to come across them when looking at a larger sample. The probability for each individual house stays the same, but the expected number increases.

The distribution starts to look more balanced with the larger sample, although it is still not close enough to a normal distribution because the expected number of these houses remains quite small.

## 5. Normal Distribution

The normal distribution is a bell-shaped curve where most values are close to the average, with fewer values at either end.

In this section, I looked at house prices to see whether they follow this type of distribution.

I used the average house price and standard deviation from the dataset to create a normal distribution curve. I then compared this curve with the actual house prices to see how closely they matched.

```python
# Calculate mean and standard deviation of house prices

mu = df['price'].mean()

sigma = df['price'].std()

print("Mean house price:", mu)
print("Standard deviation:", sigma)
```

```python
# Calculate mean and standard deviation of house prices

mu = df['price'].mean()

sigma = df['price'].std()

print("Mean house price:", mu)
print("Standard deviation:", sigma)
```

```python
from scipy.stats import norm

# Create values for the normal distribution
x_normal = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)

# Calculate the normal probability density
pdf_normal = norm.pdf(x_normal, mu, sigma)

# Plot the normal distribution
plt.plot(x_normal, pdf_normal)

plt.xlabel('House Price ($)')
plt.ylabel('Probability Density')
plt.title('Theoretical Normal Distribution of House Prices')

plt.show()
```

### Interpretation of the Normal Distribution

Looking at the graph, we can see a bell-shaped curve based on the average house price of around USD 540,089 and a standard deviation of approximately USD 367,127.

The curve is symmetrical, with its highest point around the average price. As we move further away from the average in either direction, the curve gradually gets lower.

However, this graph shows what house prices would look like if they followed a normal distribution. When I compared it with the earlier histogram, I noticed that the actual house prices were not symmetrical. Most houses were in the lower price ranges, while a few very expensive houses created a long tail on the right.

This suggests that the actual house prices do not follow a normal distribution very closely.

```python
# Compare actual house prices with the normal distribution

plt.hist(df['price'], bins=50, density=True, alpha=0.5, label='Actual House Prices')

plt.plot(x_normal, pdf_normal, color='red', label='Normal Distribution')

plt.xlabel('House Price ($)')
plt.ylabel('Probability Density')
plt.title('Actual House Prices vs Normal Distribution')

plt.legend()
plt.show()
```

### Interpretation: Actual House Prices vs Normal Distribution

Looking at the graph, we can see a clear difference between the actual house prices and the normal distribution curve.

Most houses are in the lower price ranges, while a small number of very expensive houses stretch the histogram further to the right. This shows that the actual house prices are positively skewed.

The normal distribution curve looks quite different. It has a symmetrical bell shape centred around the average house price of approximately USD 540,089.

I also noticed that the histogram has a higher peak than the normal curve. This means that more houses are grouped within certain price ranges than we would expect from a normal distribution.

Overall, the actual house prices do not follow a normal distribution very closely. The small number of extremely expensive properties is one of the main reasons for this difference.

```python
# Calculate skewness of house prices
skewness = df['price'].skew()

print("Skewness of house prices:", skewness)
```

### Interpretation of Skewness

The skewness value for house prices is around 4.02, which shows that the prices are strongly skewed to the right.

This means that most houses are in the lower and middle price ranges, while a few very expensive properties stretch the distribution further to the right.

We can also see this from the average house price, which is around USD 540,089, compared with the median of USD 450,000. The average is higher because those expensive houses pull it upwards.

A normal distribution would have a skewness value of zero. Since the value here is much higher, it supports what we noticed in the earlier graphs: house prices are not normally distributed.

## 6. Comparison of Probability Distributions

In this analysis, I looked at three probability distributions using the housing dataset: Binomial, Poisson and Normal. Each one was used to examine a different part of the data.

For the **Binomial distribution**, I looked at how many houses in a sample cost more than USD 450,000. Each house could either be above this price or not, which makes the binomial distribution suitable for this situation.

For the **Poisson distribution**, I focused on houses with seven or more bedrooms. These houses are quite rare, making up only about 0.287% of the dataset. I used this distribution to estimate how many of them we might find in a sample.

For the **Normal distribution**, I looked at the actual house prices and compared them with a normal curve. The results showed a skewness value of 4.02, which means the prices are strongly skewed to the right and do not follow a normal distribution very closely.

Overall, I found that each distribution is useful for a different purpose. The Binomial and Poisson distributions deal with counting how many times something happens, while the Normal distribution is used for continuous values such as house prices.

### Normal Approximation to the Binomial Distribution

The Binomial distribution can start to look like a Normal distribution when the sample size gets larger. A common guideline is that both np and n(1 − p) should be at least 5.

In my analysis, the probability of a house costing more than USD 450,000 was around 0.4973.

When I used a sample of 10 houses, I got np ≈ 4.97 and n(1 − p) ≈ 5.03. Since one value is just below 5, the normal approximation is not quite reliable under this guideline.

However, when I increased the sample size to 100 houses, the values increased to np ≈ 49.73 and n(1 − p) ≈ 50.27. Both are well above 5, so the normal approximation becomes much more suitable.

I could also see this difference in the graphs. With 100 houses, the Binomial distribution looked more balanced and had a shape similar to the bell-shaped Normal distribution.

This shows how increasing the sample size can make the Binomial distribution look more like a Normal distribution, even though the Binomial distribution counts whole numbers and the Normal distribution uses continuous values.

### Normal Approximation to the Poisson Distribution

The Poisson distribution can start to look like a Normal distribution when the expected number of events, known as lambda (λ), becomes large enough. A common guideline is that λ should be at least 10.

In my analysis, I looked at houses with seven or more bedrooms. These houses are quite rare, with a probability of around 0.00287.

When I used a sample of 100 houses, the lambda value was about 0.287. The graph was strongly skewed to the right, with the highest probability at zero. This means that finding no houses with seven or more bedrooms was the most likely result.

When I increased the sample size to 1,000 houses, lambda increased to around 2.869. I noticed that the distribution became more spread out and less skewed, with the highest probabilities around two or three houses.

However, 2.869 is still well below 10, so the distribution is not yet close enough to a Normal distribution for a reliable approximation.

From this comparison, I can see that increasing the sample size makes the Poisson distribution more balanced, but the expected number of rare events also needs to be large enough before a Normal approximation becomes suitable.
