### 2.1 Distribution of Housing Prices

A histogram is used to examine the frequency distribution,
spread and skewness of house prices.

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt

plt.hist(df['price'], bins=50)
plt.xlabel('House Price ($)')
plt.ylabel('Frequency')
plt.title('Histogram of Housing Prices')
plt.show()
```

Interpretation of Graph 1: Histogram of Housing Prices

The histogram shows a positively skewed distribution of housing prices. Most properties are concentrated in the lower price ranges, while a small number of expensive houses create a long right tail. The mean house price ($540,089) is higher than the median ($450,000), supporting the evidence of positive skewness. The maximum house price of $7.7 million indicates the presence of potential outliers. Therefore, the distribution of housing prices does not appear to follow a normal distribution.

### 2.2 Boxplot of Housing Prices

```python
sns.boxplot(x=df['price'])
plt.xlabel('House Price ($)')
plt.title('Boxplot of Housing Prices')
plt.show()
```

### Interpretation of Graph 2: Boxplot of Housing Prices

The boxplot illustrates the spread of housing prices and identifies potential outliers.

The median house price is USD 450,000, while the first and third quartiles are USD 321,950 and USD 645,000, respectively.

The interquartile range is USD 323,050, indicating the spread of the middle 50% of housing prices.

Numerous properties exceed the upper outlier threshold of approximately USD 1.13 million, with the maximum price reaching USD 7.7 million.

These findings support the positive skewness observed in the histogram. The extreme values may influence the mean and standard deviation, making the median and interquartile range useful measures of central tendency and variability.

### 2.3 Relationship Between Bedrooms and House Prices

```python
plt.figure(figsize=(10, 6))

sns.boxplot(x='bedrooms', y='price', data=df)

plt.xlabel('Number of Bedrooms')
plt.ylabel('House Price ($)')
plt.title('House Prices by Number of Bedrooms')

plt.show()
```

### Interpretation of Graph 3: House Prices by Number of Bedrooms

The boxplot compares housing prices across properties with different numbers of bedrooms.

The results suggest that median house prices generally increase as the number of bedrooms increases, particularly for properties with two to seven bedrooms.

Properties with four, five and six bedrooms show considerable price variation and numerous high-priced outliers. Some properties exceed USD 7 million.

The dataset also contains an unusual observation of 33 bedrooms, which may represent a data-entry error or an exceptional property.

Overall, the graph suggests an association between bedroom count and house price. However, other factors such as location, property size and condition may also influence housing prices.
